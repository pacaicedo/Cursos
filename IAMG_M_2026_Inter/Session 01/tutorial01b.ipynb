##Introduction
The Wine Quality dataset, which is a popular dataset for regression tasks. It contains various chemical properties of wines and their quality rating. However, we can adapt it to a classification problem by categorising wine quality into classes.

In this tutorial, we will utilise modified version of the dataset, which is already split into training and test subsets.

The aim with this tutorial is to implement a Dense Neural Network model that can predict whether a given wine is good or bad quality.

Let's import the required libraries:

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

## Data preparation
First, let's mount the Google Drive so we can access to the required datasets

* Load the training and test splits, which are named `winequality-red_train.csv` and `winequality-red_test.csv`, respectively.

```python
# Load the datasets. Don't forget to change the below path with your own one.

X_train = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/2025SUM/Session 01/winequality-red_train.csv', index_col=0)
X_test = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/2025SUM/Session 01/winequality-red_test.csv', index_col=0)

X_train.head()
```

* Similarly, load the target data:

```python
Y_train = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/2025SUM/Session 01/winequality-red_train_labels.csv', index_col=0)
Y_test = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/2025SUM/Session 01/winequality-red_test_labels.csv', index_col=0)

Y_train.head()
```

* Neural network algorithms required the features (input variables) to be normalised (or scaled):

```python
# Standardize the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

##Vanilla model

We will build a vanila model with a dense neural network architecture as follows:

```python
model = Sequential([
    Dense(128, input_dim=X_train.shape[1], activation='relu'),  # Input layer
    Dropout(0.5),                                               # Dropout for regularization
    Dense(64, activation='relu'),                               # Hidden layer
    Dropout(0.5),                                               # Dropout for regularization
    Dense(1, activation='sigmoid')                              # Output layer
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['AUC'])                                  # Using AUC as metric

model.summary()
```

... and train the model using `X_train` and `Y_train`:

```python
model.fit(X_train, Y_train, epochs=20, batch_size=32, validation_split=0.2)
```

```python
test_loss, test_auc = model.evaluate(X_test, Y_test)

print('\nTest AUC:', test_auc)
```
