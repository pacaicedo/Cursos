# Descripción utilizando grafos de la enfermedad cardiovascular

## Cargue de librerías

```python
import kagglehub

import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Configuración de script

```python
print("Project Path:", Path.cwd())
PROJECT_ROOT = Path.cwd() / "MGCD_M_2026_Inter/little_exercise/"

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_DIR = PROJECT_ROOT / "data"
LOCAL_FILE = DATA_DIR / "cardio_train.csv"

DATASET_HANDLE = "sulianova/cardiovascular-disease-dataset"
DATASET_FILE = "cardio_train.csv"
```

```python
path = kagglehub.dataset_download(DATASET_HANDLE, output_dir=str(DATA_DIR))
print("Path to dataset files:", path)
```

## Cargue de datos

```python
if LOCAL_FILE.exists():
    print(f"El archivo ya existe. No se descarga nada: {LOCAL_FILE}")
    data = pd.read_csv(LOCAL_FILE, sep=";")
else:
    downloaded_file = kagglehub.dataset_download(DATASET_HANDLE,output_dir=str(DATA_DIR))
    print(f"El archivo no existe. Descargando desde Kaggle... {downloaded_file}")
    data = pd.read_csv(downloaded_file, sep=";")
```
